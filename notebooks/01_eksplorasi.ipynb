```python
import sys
import pandas as pd
sys.path.append("../backend")
from nlp_utils import clean_text

df = pd.read_excel("../data/dataset_chatbot_kesehatan.xlsx", sheet_name="Dataset")
df["pertanyaan"] = df["pertanyaan"].astype(str).str.strip()
df.head()
```

```python
print(df.shape)
print(df.isna().sum())
print(df["intent"].value_counts())
print("Pertanyaan duplikat:", df["pertanyaan"].duplicated().sum())
print(df.groupby("intent")["jawaban"].nunique())
```

```python
df["clean"] = df["pertanyaan"].apply(clean_text)
print(df[["pertanyaan", "clean"]].sample(5, random_state=1).to_string())
print("Panjang kalimat maksimum (kata):", df["clean"].str.split().str.len().max())
```
